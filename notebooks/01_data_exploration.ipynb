```python
import pandas as pd

path = "../data/benchmark/snapshots/kg_t0.csv"

sample = pd.read_csv(
    path,
    nrows=100_000
)

print("Sample shape:", sample.shape)

print("\nX entity types:")
print(sample["x_type"].value_counts())

print("\nY entity types:")
print(sample["y_type"].value_counts())

print("\nTop relations:")
print(sample["relation"].value_counts().head(30))
```

```python
print("\nDrug-related rows:")
print(
    sample[
        (sample["x_type"] == "drug") |
        (sample["y_type"] == "drug")
    ][
        ["x_type", "x_name", "relation", "y_type", "y_name"]
    ].head(20)
)

print("\nDisease-related rows:")
print(
    sample[
        (sample["x_type"] == "disease") |
        (sample["y_type"] == "disease")
    ][
        ["x_type", "x_name", "relation", "y_type", "y_name"]
    ].head(20)
)
```

```python
import pandas as pd

path = "../data/benchmark/snapshots/kg_t0.csv"

# Read the file in chunks and take a sample from different parts
chunks = []

for chunk in pd.read_csv(path, chunksize=100_000):
    chunks.append(chunk.sample(min(1_000, len(chunk)), random_state=42))

sample = pd.concat(chunks, ignore_index=True)

print("Sample shape:", sample.shape)

print("\nX entity types:")
print(sample["x_type"].value_counts())

print("\nY entity types:")
print(sample["y_type"].value_counts())

print("\nTop relations:")
print(sample["relation"].value_counts().head(30))
```
